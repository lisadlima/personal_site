1. **Discovery**: A business (merchant) publishes a JSON profile at `/.well-known/ucp` on their domain. It lists their services, capabilities, payment handlers, and signing keys.

4. **Transports**: REST, MCP, A2A, or Embedded. MCP uses `tools/call` with the operation name — so a UCP merchant's MCP server would expose tools like `create_checkout`, `search_catalog`, etc.

5. **Payment**: Decoupled architecture — the business advertises payment handlers, the platform acquires tokens from the payment provider, then sends the token to the business to complete. For autonomous agents there's an AP2 mandate extension with cryptographic proof.

**What we have right now:** our mock merchant's `complete_checkout` just accepts a `payment_token` string — it's a stand-in, no real payment processing.

**What UCP actually specifies for payment:** it's a decoupled, three-party model:

Good instinct to bring in A2A — it's the piece that makes the "agents talking to each other" story real, and it's directly relevant to NVIDIA's stack.

**A2A (Agent-to-Agent) in a nutshell:**

Where MCP connects a *model to tools* (vertical: agent reaches down to resources), A2A connects *agents to each other* (horizontal: peer agents that treat each other as opaque black boxes). You don't see the other agent's internal tools or model — you just exchange work. The core concepts:

- **Agent Card**: a JSON document published at a well-known URL (like `/.well-known/agent.json`) advertising an agent's identity, skills, endpoint, and auth requirements. This is the discovery layer — very analogous to UCP's profile.
- **Task**: the unit of work. A client agent sends a task to a remote agent, which works on it and returns results. Tasks have a lifecycle — `submitted`, `working`, `input-required` (it needs more from you), `completed`, `failed`.
- **Messages & Parts**: the back-and-forth within a task, made of typed parts (text, file, structured data) — the same shape as the `Part` objects we're already using with fastllm.
- **Artifacts**: the concrete outputs the remote agent produces.
- **Streaming**: long-running tasks stream progress over SSE.

Under the hood it's HTTP + JSON-RPC, so structurally very similar to what solvemcp already does.

**How it fits our demo:** A2A is actually one of UCP's listed transports. So there are two natural places it shows up:
1. **Between our own agents** — Requirements → Sourcing → Procurement could hand off tasks to each other as A2A peers, rather than being one loop calling functions.
2. **Buyer ↔ merchant** — instead of the buyer hitting REST endpoints, the buyer agent could send a task to the merchant's *agent* (which has its own Agent Card).

`mistletoe` is a Markdown-to-HTML converter (a pure-Python library, similar in spirit to `markdown` or `mistune`). We need it because our agents return their output as **Markdown** — headings, tables, bullet lists, `**bold**`, etc. But a browser doesn't render Markdown; it renders HTML. So `mistletoe.markdown(text)` turns that Markdown string into an HTML string the browser can display properly.

Without it, your sourcing report's nice tables and headings would show up as literal `##` and `|---|` characters on the page instead of a formatted table.

The one extra piece: in `Panel` we wrap it in `Div(..., cls='marked')`. That `marked` class ties into fasthtml/PicoCSS styling so the rendered HTML looks clean. The raw HTML from mistletoe gets inserted into the div as the panel body.

So the flow is: agent returns Markdown → `mistletoe.markdown()` → HTML → shown in a card on the page.

Kind of, but not quite. **A2A is less like a group chat and more like one agent hiring another agent to complete a task.**

In your staged pipeline, the orchestrator does this:

```text
Python orchestrator → Requirements Agent
Python orchestrator → Sourcing Agent
Python orchestrator → Procurement Agent
```

The agents do not know much about each other. They just receive text and return text.

With **A2A**, each agent is more like a network service with an identity and advertised capabilities:

```text
Requirements Agent publishes an Agent Card
Sourcing Agent publishes an Agent Card
Procurement Agent publishes an Agent Card
```

An **Agent Card** says things like:

```text
I am the Sourcing Agent.
I can discover suppliers.
I can search UCP catalogs.
Send tasks to me at this endpoint.
Here is how auth works.
```

Then another agent can send it a **task**:

```text
Task: Find qualified suppliers for these constraints.
```

The remote agent works on that task and returns artifacts/results.

So the difference is:

| Staged orchestration | A2A |
|---|---|
| One Python function coordinates everything | Agents expose network endpoints |
| Outputs passed directly to next function | Agents send tasks/messages to each other |
| Simple and notebook-friendly | More production-like and distributed |
| Great for v1 demo | Great for scaling across teams/orgs |
| Agent boundaries are conceptual | Agent boundaries are protocol-level |

```text
Buyer Agent → Requirements Agent: "Extract constraints"
Buyer Agent → Sourcing Agent: "Find qualified products"
Buyer Agent → Procurement Agent: "Prepare checkout"
```

```python
def search_products(query:str="", filters:dict=None):
```
